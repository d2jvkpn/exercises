```python
import numpy as np
```

```python
def sliding_windows(tokens, block_size, stride):
    # tokens: 1D np.array of int (length L)
    L = len(tokens)
    if L < block_size:
        return []

    starts = np.arange(0, L - block_size + 1, stride)
    return [tokens[s:s+block_size] for s in starts]

def chunk_windows(tokens, block_size):
    L = len(tokens)
    n = L // block_size
    return [tokens[i*block_size:(i+1)*block_size] for i in range(n)]
```

```python
L, ctx, stride = 10000, 2048, 256
tokens = np.arange(L)
```

```python
windows = sliding_windows(tokens, ctx, stride)  # 形如 [ (2048,), (2048,), ... ]

print("=> sliding_windows: [:5]")
for i in range(5):
    print(windows[i])
```

```python
windows = chunk_windows(tokens, ctx)

print("=> chunk_windows:")
for i in range(len(windows)):
    print(windows[i])
```
